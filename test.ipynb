# ETL Pipeline for Current Weather in AU

# 1.0 Imports

```python
import os
import json
import requests
import pandas as pd
from mysql import connector
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
API_KEY = os.getenv('API_KEY')
API_HOST = os.getenv('API_HOST')
cities_chosen = ['Sydney','Melbourne','Perth','Brisbane','Adelaide','Point Cook','Gold Coast', 'Newcastle', 'Wollongong']
```

```python
url = 'https://open-weather13.p.rapidapi.com/city'
headers = {
    'x-rapidapi-key': API_KEY,
    'x-rapidapi-host': API_HOST
}
```

# 2.0 Extract

```python
all_weather_data = []
for cities in cities_chosen:
    query_string = {
        'lang' : 'EN',
        'city' : cities
    }
    response = requests.get(url=url,
                            headers=headers,
                            params=query_string)

    if response.status_code == 200:
        payload = response.json()
        city_record = {
            "city_name": payload["name"],
            "temperature": payload["main"]["temp"],
            "humidity": payload["main"]["humidity"],
            "description": payload["weather"][0]["description"],
            "lat": payload["coord"]["lat"],
            "lon": payload["coord"]["lon"]
        }
        all_weather_data.append(city_record)
        print(f"[SUCCESS] Data Received for {cities}!")
    else:
        print(f"[ERROR] Connection Unsuccessful! for {cities}!")
```
